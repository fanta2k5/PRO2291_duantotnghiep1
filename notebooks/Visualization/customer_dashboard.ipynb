## BIỂU ĐỒ SỐ KHÁCH HÀNG THEO GIỚI TÍNH (Customer by Gender)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

gender_count = (
    df.groupby("gender")["cust_id"]
      .nunique()
      .sort_values(ascending=False)
)

plt.figure(figsize=(7,5))

gender_count.plot(kind="bar")

plt.title("Customer Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)

plt.show()
```

## BIỂU ĐÔ PHÂN BỐ KHÁCH HÀNG THEO ĐỘ TUỔI (Age Distribution)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

plt.figure(figsize=(10,6))

plt.hist(
    df["age"].dropna(),
    bins=10
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

## BIỂU ĐỒ DOANH THU THEO GIỚI TÍNH (Revenue by Gender)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

gender_revenue = (
    df.groupby("gender")["total"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

bars = plt.bar(
    gender_revenue.index,
    gender_revenue.values
)

plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue")

# Hiển thị giá trị trên cột
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

## BIỂU ĐỒ DOANH THU THEO NHÓM TUỔI (Revenue by Age Group)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 25, 35, 45, 60, 100],
    labels=["0-18", "19-25", "26-35", "36-45", "46-60", "60+"]
)

age_count = (
    df.groupby("age_group")
      .size()
      .reindex(["0-18", "19-25", "26-35", "36-45", "46-60", "60+"])
)

plt.figure(figsize=(8,5))
age_count.plot(kind="bar")

plt.title("Number of Customers by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

plt.show()
```

## BIỂU ĐÒ TOP 10 KHÁCH HÀNG CÓ DOANH THU CAO NHẤT (Top 10 Customers)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv")

top_customer = (
    df.groupby("cust_id")["total"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customer.plot(kind="bar")

plt.title("Top 10 Customers by Revenue")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```

## Dashboard Customer

```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu
df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

# Customer by Gender
gender_count = (
    df.groupby("gender")["cust_id"]
      .nunique()
)

# Revenue by Gender
gender_revenue = (
    df.groupby("gender")["total"]
      .sum()
)

# Age Group
df["age_group"] = pd.cut(
    df["age"],
    bins=[0,18,25,35,45,60,100],
    labels=["0-18","19-25","26-35","36-45","46-60","60+"]
)

age_count = (
    df.groupby("age_group")
      .size()
      .reindex(["0-18","19-25","26-35","36-45","46-60","60+"])
)

# Top Customers
top_customer = (
    df.groupby("cust_id")["total"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)


# Dashboard
fig = plt.figure(figsize=(18,10))

fig.suptitle(
    "Customer Analytics Dashboard",
    fontsize=22,
    fontweight="bold"
)

# Customer by Gender
ax1 = plt.subplot(2,3,1)

gender_count.plot(
    kind="bar",
    ax=ax1
)

ax1.set_title("Customer by Gender")
ax1.set_xlabel("Gender")
ax1.set_ylabel("Customers")

# Revenue by Gender
ax2 = plt.subplot(2,3,2)

bars = ax2.bar(
    gender_revenue.index,
    gender_revenue.values
)

ax2.set_title("Revenue by Gender")
ax2.set_xlabel("Gender")
ax2.set_ylabel("Revenue")

for bar in bars:
    ax2.text(
        bar.get_x()+bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        fontsize=9
    )

# Age Distribution
ax3 = plt.subplot(2,3,3)

ax3.hist(
    df["age"].dropna(),
    bins=10
)

ax3.set_title("Age Distribution")
ax3.set_xlabel("Age")
ax3.set_ylabel("Customers")


# Customer by Age Group
ax4 = plt.subplot(2,3,4)

age_count.plot(
    kind="bar",
    ax=ax4
)

ax4.set_title("Customer by Age Group")
ax4.set_xlabel("Age Group")
ax4.set_ylabel("Customers")

# Top 10 Customers
ax5 = plt.subplot(2,3,(5,6))

top_customer.plot(
    kind="bar",
    ax=ax5
)

ax5.set_title("Top 10 Customers by Revenue")
ax5.set_xlabel("Customer ID")
ax5.set_ylabel("Revenue")

plt.tight_layout()

plt.show()
```
