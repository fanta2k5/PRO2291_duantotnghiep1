## BIỂU ĐỒ SỐ KHÁCH HÀNG THEO GIỚI TÍNH (Customer by Gender)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

gender_count = (
    df.groupby("gender")["cust_id"]
      .nunique()
      .sort_values(ascending=False)
)

plt.figure(figsize=(7,5))

gender_count.plot(kind="bar")

plt.title("Customer Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)

plt.show()
```

## BIỂU ĐÔ PHÂN BỐ KHÁCH HÀNG THEO ĐỘ TUỔI (Age Distribution)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

plt.figure(figsize=(10,6))

plt.hist(
    df["age"].dropna(),
    bins=10
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

## BIỂU ĐỒ DOANH THU THEO GIỚI TÍNH (Revenue by Gender)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

gender_revenue = (
    df.groupby("gender")["total"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

bars = plt.bar(
    gender_revenue.index,
    gender_revenue.values
)

plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue")

# Hiển thị giá trị trên cột
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

## BIỂU ĐỒ DOANH THU THEO NHÓM TUỔI (Revenue by Age Group)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv"
)

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 25, 35, 45, 60, 100],
    labels=["0-18", "19-25", "26-35", "36-45", "46-60", "60+"]
)

age_count = (
    df.groupby("age_group")
      .size()
      .reindex(["0-18", "19-25", "26-35", "36-45", "46-60", "60+"])
)

plt.figure(figsize=(8,5))
age_count.plot(kind="bar")

plt.title("Number of Customers by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

plt.show()
```

## BIỂU ĐÒ TOP 10 KHÁCH HÀNG CÓ DOANH THU CAO NHẤT (Top 10 Customers)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\PRO2291_DuAnTopNghiep\PRO2291_duantotnghiep1\data\cleaned\sales_cleaned.csv")

top_customer = (
    df.groupby("cust_id")["total"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customer.plot(kind="bar")

plt.title("Top 10 Customers by Revenue")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```
