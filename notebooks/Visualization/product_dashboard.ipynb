```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

df = pd.read_excel("../../data/cleaned/sales_cleaned.xlsx")

df.head()
```

# Product Dashboard

Dashboard phân tích doanh thu và hiệu quả kinh doanh theo sản phẩm nhằm xác định danh mục và sản phẩm mang lại doanh thu cao cho doanh nghiệp.


```python
category = (
    df.groupby("category")["total"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=category.values,
    y=category.index
)

plt.title("Revenue by Category")
plt.xlabel("Revenue")
plt.ylabel("Category")

plt.show()
```

```python
qty = (
    df.groupby("category")["qty_ordered"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=qty.values,
    y=qty.index
)

plt.title("Quantity Sold by Category")
plt.xlabel("Quantity")
plt.ylabel("Category")

plt.show()
```

```python
top10 = (
    df.groupby("sku")["total"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top10.values,
    y=top10.index
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("SKU")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["total"],
    bins=30,
    kde=True
)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df["price"]
)

plt.title("Price Distribution")

plt.show()
```
